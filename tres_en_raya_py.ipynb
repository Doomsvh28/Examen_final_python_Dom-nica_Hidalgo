```python
#Doménica Hidalgo
#Examen final
#14.05.2026

import random
class TresEnRaya:
  def __init__(self, nombre_j1, nombre_j2):

    self.jugadores= {"x": nombre_j1, "O": nombre_j2}
    self.tablero = [
        [" "," "," "],
        [" "," "," "],
        [" "," "," "]
    ]
    self.jugador1 = nombre_j1
    self.jugador2= nombre_j2
    self.simboloj1 = "X" or "O"
    self.simboloj2 = "X" or "O"


  def mostrar_tablero(self, con_numeros=True):
        print("\n")
        cont = 1
        for fila in range(3):
            linea = ""
            for col in range(3):
                contenido = self.tablero[fila][col]

                if contenido == " " and con_numeros:
                    linea += f" {cont} "
                else:
                    linea += f" {contenido} "

                if col < 2: linea += "|"
                cont += 1
            print(linea)
            if fila < 2: print("---" * 2 + "---")
        print("\n")



  def colocar_ficha(self,posicion):
    if not (1 <= posicion <= 9):
      return False


  def verificar_ganador(self):
    for i in range(3):
      if self.tablero[i][0] == self.tablero[i][1] == self.tablero[i][2] != " ":
        return self.tablero[i][0]
      if self.tablero[0][i] == self.tablero[1][i] == self.tablero[2][i] != " ":
        return self.tablero[0][i]


  def reinicio(self):
        self.tablero = [[" " for _ in range(3)] for _ in range(3)]
        self.jugador_actual = "X"
        self.estado_activo = True


  def tablero_lleno(self):
        return all(casilla != " " for fila in self.tablero for casilla in fila)


  historial = {
      "nombre": "",
      "victorias": "",
      }

print("== TRES EN RAYA ==")
print("Cargando historial...")
j1 = input("Nombre del Jugador 1 (X): ")
j2 = input("Nombre del Jugador 2 (O): ")

juego = TresEnRaya(j1, j2)
juego = random.randint({j1, j2})
print(f"Incia {juego}")
while True:
        juego.reinicio()
        while True:
            juego.mostrar_tablero()
            print(f"Turno de {juego.jugadores[juego.jugador_actual]} ({juego.jugador_actual})")
pos = int(input("Elige una posición (1-9): "))
if pos > 9 or pos < 0:
  print("¡Movimiento inválido! Intenta de nuevo.")

```
